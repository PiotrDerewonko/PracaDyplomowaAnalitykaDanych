Sprawdzenie czy kolumny liczbowe nie zawierają ujemnych wartości

```python
list_of_numeric_columns=['NUMBER OF PERSONS INJURED', 'NUMBER OF PERSONS KILLED', 'NUMBER OF PEDESTRIANS INJURED', 'NUMBER OF PEDESTRIANS KILLED', 'NUMBER OF CYCLIST INJURED', 'NUMBER OF CYCLIST KILLED', 'NUMBER OF MOTORIST INJURED', 'NUMBER OF MOTORIST KILLED' ]
```

```python
for i in list_of_numeric_columns:
    min_value = data[i].min()
    if min_value < 0:
        print(f'kolumna {i} zawiera ujemne wartości, prawdopodobnie błędne dane')
    else:
        print(f'kolumna {i} zawiera zerowe lub dodatnie wartości')
```
