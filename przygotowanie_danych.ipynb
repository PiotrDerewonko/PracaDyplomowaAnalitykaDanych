Sprawdzenie czy kolumny liczbowe o zabitych i rannych nie zawierają ujemnych wartości

```python
list_of_numeric_columns_injured_or_killed=['NUMBER OF PERSONS INJURED', 'NUMBER OF PERSONS KILLED', 'NUMBER OF PEDESTRIANS INJURED', 'NUMBER OF PEDESTRIANS KILLED', 'NUMBER OF CYCLIST INJURED', 'NUMBER OF CYCLIST KILLED', 'NUMBER OF MOTORIST INJURED', 'NUMBER OF MOTORIST KILLED']
```

```python
for i in list_of_numeric_columns_injured_or_killed:
    min_value = data[i].min()
    if min_value < 0:
        print(f'kolumna {i} zawiera ujemne wartości, prawdopodobnie błędne dane')
    else:
        print(f'kolumna {i} zawiera zerowe lub dodatnie wartości')
```

Sprawdzenie czy suma kolumn o zabitych i rannych jest zgodna z kolumnami z sumami

```python
dict_killed_injured = {'NUMBER OF PERSONS INJURED': [['NUMBER OF PEDESTRIANS INJURED', 'NUMBER OF CYCLIST INJURED', 'NUMBER OF MOTORIST INJURED' ], 'CHECK_INJURED'], 'NUMBER OF PERSONS KILLED': [['NUMBER OF PEDESTRIANS KILLED', 'NUMBER OF CYCLIST KILLED', 'NUMBER OF MOTORIST KILLED'], 'CHECK_KILLED']}
data[['CHECK_KILLED', 'CHECK_INJURED']] = 0
for i, j in dict_killed_injured.items():
    for k in j[0]:
        data[j[1]] = data[j[1]] + data[k]
data
```

Ze względu na niewielką ilość błędnych danych o sumie zabitych lub rannych usuwam te dane.

```python
print(f'Ilośc wierszy przed usunięciem: {len(data)}')
data = data.drop(data.loc[data['NUMBER OF PERSONS INJURED'] != data['CHECK_INJURED'] ].index)
data = data.drop(data.loc[data['NUMBER OF PERSONS KILLED'] != data['CHECK_KILLED'] ].index)
print(f'Ilośc wierszy po usunięciu: {len(data)}')
data = data.drop(columns=['CHECK_INJURED', 'CHECK_KILLED'])
```

Sprawdzenie czy kolumna COLLISION_ID zawiera unikatowe wartośc

```python
data['COLLISION_ID'].is_unique
```

kolumna COLLISION_ID zawiera duplikaty, dlatego sprawdzam które rekordy zawierają duplikaty

```python
data['tmp'] = 1
check_collision_id = data.pivot_table(index='COLLISION_ID', values='tmp', aggfunc='count').sort_values(by='tmp', ascending=False)
duplicate_collision_id = check_collision_id.loc[check_collision_id['tmp'] > 1]
```

sprawdzam czy zdublowane collision_id zawierają dokładnie te same informacje

```python
bad_groups = data.groupby("COLLISION_ID").nunique().gt(1).any(axis=1).sort_values(ascending=False)
print(f'jest {len(bad_groups[bad_groups])} wierszy które zawierają rózne dane w ramach tego samogo collision_id')
print(f'jest {len(bad_groups[~bad_groups])} wierszy które zawierają takie same dane w ramach tego samogo collision_id')
```

w związku z tym, że dane w ramach collision id zawierają dokłądnie te same dane usuwam zdublowane rekordy

```python
print(f'ilośc wierszy przed usunięciem duplikatów: {len(data)}')
data = data.drop_duplicates()
print(f'ilośc wierszy po usunięciu duplikatów: {len(data)}')
```
